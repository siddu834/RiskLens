```python
import sys
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
# Set the project root directory
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Allow Python to import modules from src/
sys.path.insert(0, str(PROJECT_ROOT))
from src.config import (
    RANDOM_SEED,
    TEST_SIZE,
    TARGET_COL
)

# Load original dataset and remove Kaggle index column
df = pd.read_csv(PROJECT_ROOT / "data" / "cs-training.csv", index_col=0)

print("Original dataset shape:", df.shape)

# Create stratified 80:20 split
dev, test = train_test_split(
    df,
    test_size=TEST_SIZE,
    stratify=df[TARGET_COL],
    random_state=RANDOM_SEED
)

# Save without additional index columns
dev.to_csv(PROJECT_ROOT / "data" / "dev.csv", index=False)
test.to_csv(PROJECT_ROOT / "data" / "test.csv", index=False)

print("Dev shape:", dev.shape)
print("Test shape:", test.shape)

print("Dev default rate:", dev[TARGET_COL].mean())
print("Test default rate:", test[TARGET_COL].mean())
```
